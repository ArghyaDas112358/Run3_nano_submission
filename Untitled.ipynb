```python
import ROOT

fname = "/eos/purdue/store/user/arghyara/production/Scouting/NanoAODv15Scouting24/mcscouting_2024/DYto2L-2Jets_Bin-MLL-50-PTLL-100_TuneCP5_13p6TeV_amcatnloFXFX-pythia8/RunIII2024Summer24MiniAODv6-150X_mcRun3_2024_realistic_v2-v3_DAZSLE_PFNano/260203_185118/0000/MC_2024_Scouting_1.root"

ROOT.gROOT.SetBatch(False)

ROOT.EnableImplicitMT()

df = ROOT.RDataFrame("Events", fname)
```

```python
f = ROOT.TFile.Open(fname)
t = f.Get("Events")
brs = [b.GetName() for b in t.GetListOfBranches()]

for k in ["genWeight", "Generator_weight", "LHEWeight_originalXWGTUP",
          "puWeight", "Pileup_nTrueInt", "L1PreFiringWeight_Nom",
          "XSWeight", "eventWeight", "weight"]:
    if k in brs:
        print("found:", k)
```

```python
from IPython.display import Image, display, clear_output

df_mu = (
    df.Filter("nScoutingMuonVtx > 0")
      .Define("idx_mu", "ROOT::VecOps::Argsort(ScoutingMuonVtx_pt)")  # ascending
      .Define("i1", "idx_mu[idx_mu.size()-1]")                       # leading
      .Define("lead_mu_pt", "ScoutingMuonVtx_pt[i1]")
      .Define("w", "genWeight")
)

h_mu1 = df_mu.Histo1D(
    ("h_leadmu", "Leading ScoutingMuonVtx p_{T};p_{T} [GeV];Events", 80, 0, 400),
    "lead_mu_pt", "w"
)


if "c_mu" not in globals(): c_mu = ROOT.TCanvas("c_mu", "Leading scouting muon pT", 800, 600)
c_mu.cd()
c_mu.Clear()
h_mu1.Draw("hist")
c_mu.Modified()

out_png = "lead_mu_pt.png" 
c_mu.SaveAs(out_png)
clear_output(wait=True)
display(Image(filename=out_png))
```

```python
df_el = (
    df.Filter("nScoutingElectron > 0")
      .Define("idx_el", "ROOT::VecOps::Argsort(ScoutingElectron_pt)")  # ascending
      .Define("j1", "idx_el[idx_el.size()-1]")                         # leading
      .Define("lead_el_pt", "ScoutingElectron_pt[j1]")
      .Define("w", "genWeight")
)

h_el1 = df_el.Histo1D(
    ("h_leadel", "Leading ScoutingElectron p_{T};p_{T} [GeV];Events", 80, 0, 200),
    "lead_el_pt", "w"
)


if "c_el" not in globals(): c_el = ROOT.TCanvas("c_el", "Leading scouting muon pT", 800, 600)
c_el.cd()
h_el1.Draw("hist")
c_el.Modified()

out_png = "lead_el_pt.png" 
c_el.SaveAs(out_png)
clear_output(wait=True)
display(Image(filename=out_png))
```

```python
df_mumu = (
    df.Filter("nScoutingMuonVtx >= 2")
      # sort by pT
      .Define("idx_mu", "ROOT::VecOps::Argsort(ScoutingMuonVtx_pt)")   # ascending
      .Define("i1", "idx_mu[idx_mu.size()-1]")                        # leading
      .Define("i2", "idx_mu[idx_mu.size()-2]")                        # subleading

      # optional but strongly recommended: OS requirement
      .Filter("ScoutingMuonVtx_charge[i1] * ScoutingMuonVtx_charge[i2] < 0")

      # build 4-vectors (muon mass in GeV)
      .Define(
          "p4_1",
          "ROOT::Math::PtEtaPhiMVector("
          "ScoutingMuonVtx_pt[i1], "
          "ScoutingMuonVtx_eta[i1], "
          "ScoutingMuonVtx_phi[i1], "
          "0.105658)"
      )
      .Define(
          "p4_2",
          "ROOT::Math::PtEtaPhiMVector("
          "ScoutingMuonVtx_pt[i2], "
          "ScoutingMuonVtx_eta[i2], "
          "ScoutingMuonVtx_phi[i2], "
          "0.105658)"
      )

      # dilepton pT
      .Define("mumu_pt", "(p4_1 + p4_2).Pt()")
      .Define("w", "genWeight")
)

h_mumu = df_mumu.Histo1D(
    ("h_mumu_pt",
     "Reco di-#mu p_{T};p_{T}(#mu#mu) [GeV];Events",
     120, 0, 600),
    "mumu_pt", "w"
)


if "c_mumu" not in globals():
    c_mumu = ROOT.TCanvas("c_mumu", "Reco di-muon pT", 800, 600)

c_mumu.cd()
c_mumu.Clear()
h_mumu.Draw("hist")
c_mumu.Modified()

out_png = "mumu_pt.png"
c_mumu.SaveAs(out_png)

clear_output(wait=True)
display(Image(filename=out_png))

```
